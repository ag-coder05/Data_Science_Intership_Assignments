```python
import pandas as pd
import numpy as np
df= pd.read_csv('StudentsPerformance.csv')
sub = ['math score', 'reading score', 'writing score']
print('Student Report')
for s in sub:
  #task 1: clean data by filling any NaN with mean of that score
  m= df[s].mean()
  df[s]= df[s].fillna(m)

  #task2: calculate average, max and min
  avg=df[s].mean()
  mx=df[s].max()
  mn=df[s].min()

  #task3: display result
  print(f"Subject: {s}")
  print(f"  Average Score: {avg}")
  print(f"  Highest Score: {mx}")
  print(f"  Lowest Score:  {mn}")
```
